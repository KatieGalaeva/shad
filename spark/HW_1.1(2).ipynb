**2.1 Решаем задачу подсчета слов в тексте**

В рамках данного задания вам нужно подсчитать кол-во слов в тексте Шекспира, по сути необходимо реализовать базовый функционал утилиты word count.

**Важно**, подсчитывайте число только тех слов, длина которых больше 4 символов. Проводить процесс удаления знаков препинания и прочих символов **не нужно**

Ниже вам предоставлены ячейки, в которые вы можете засипать структуру маппера/редьсюера, и ниже вызвать их в общем скрипта запуска map reduce таски

````python
%%writefile mapper.py

import sys


for line in sys.stdin:
    words = line.strip().split(" ")
    for word in words:
        if len(word) > 4:
            print(word.lower(), 1, sep="+")

````

````python
%%writefile reducer.py

import sys

current_word = None
word_count = 0

for line in sys.stdin:
    word, count = line.strip().split('+')  
    count = int(count)

    if current_word != word:
        if current_word:
            print(current_word, word_count, sep='\t')
        current_word = word
        word_count = 0

    word_count += count

if current_word:
    print(current_word, word_count, sep='\t')
````

Также как и в задании 2.1, вы можете создать набор ячеек, куда занесете код своих маппера и редьюсера, и также соберете код запуска MR таски. Нам нужно, чтобы вы предоставили нам в качестве решения этой задачи все эти составляющие в рамках этого ноутбука.

````python
%%writefile mapper.py

import sys

for line in sys.stdin:
    line = line.strip()
    if not line or line.startswith("country"):  
        continue
    try:
        country, company, date, income = line.split(";")
        print(f"{country}+{company}\t{income}")
    except ValueError:
        continue
    
````

````python
%%writefile reducer.py


import sys

current_key = None
max_income = 0

for line in sys.stdin:
    line = line.strip()
    if not line:
        continue
    try:
        key, income = line.split("\t")
        income = int(income)
    except ValueError:
        continue

    if current_key != key:
        if key is not None:
            print(f"{current_key}\t{max_income}")
        current_key = key
        max_income = income
    else:
        max_income = max(max_income, income)
        
if current_key is not None:
    print(f"{current_key}\t{max_income}")

````

**2.3 Решаем задачу поиска самых посещаемых сайтов**

В данном задании нужно поработать с логом данных о посещении юзерами различных сайтов.
Формат данных: `url;временная метка`. Вам нужно вывести топ 5 сайтов по посещаемости в каждую из дат, которая представлена в наших данных.

Результат работы скрипта должен выглядеть следующим образом:

```bash

2024-05-25  https://gonzales-bautista.com/  987
```

Также в рамках этой задачи вам могу пригодиться дополнительные параметры mr таски, отвечающие за настройку шаффла, и правил сортировки ключей внутри него. Почитать о примерах их использования можно [здесь](https://hadoop.apache.org/docs/current/hadoop-streaming/HadoopStreaming.html#More_Usage_Examples).

Также не рекомендуем использовать `\t` в качестве символа разделителя для сложного ключа (потому что по дефолту таб используется для разделения колонок данных, и ключом в таком случае будет только первая колонка до таба). Если вы будете собирать сложный ключ для нужной вам сортировки данных, лучше всего будет использовать другие симловы, к примеру `+, =`.

Возможно, у вас не получится решить данную задачу за одну mr таску, тогда вы просто описываете в решении скрипты ваших мапперов, редьюсеров под каждую из mr тасок, которые вам нужно запустить для получения нужного результата.

**Важно** помнить, что любой маппер и редьюсер должен работать за O(1) памяти, и если вы будете создавать какой-то список, куда будете складывать какие-то данные, то он не должен быть размера O(n). Если такой момент в вашем решении будет, пожалуйста, поясните его текстово, что с вашими переменными все ок, и у них нет размера O(n). 

````python
%%writefile mapper1.py


import sys

for line in sys.stdin:
    line = line.strip()
    if not line or ";" not in line:
        continue
    try:
        url, timestamp = line.split(";", 1)  
        date = timestamp.strip().split(" ")[0]  
        print(f"{date}+{url}\t1")
    except Exception:
        continue

````

````python
%%writefile reducer1.py

import sys

current_key = None
total = 0

for line in sys.stdin:
    line = line.strip()
    try:
        key, count = line.split("\t")
        count = int(count)
    except ValueError:
        continue

    if key != current_key:
        if current_key:
            print(f"{current_key}\t{total}")
        current_key = key
        total = count
    else:
        total += count

if current_key:
    print(f"{current_key}\t{total}")
````

````python
%%writefile mapper2.py
import sys

for line in sys.stdin:
    line = line.strip()
    try:
        key, count = line.split("\t")
        date, url = key.split("+", 1)
        print(f"{date}\t{url}+{count}")
    except ValueError:
        continue
````

````python
%%writefile reducer2.py

import sys
from heapq import nlargest

current_date = None
entries = []

for line in sys.stdin:
    line = line.strip()
    try:
        date, rest = line.split("\t")
        url, count = rest.rsplit("+", 1)
        count = int(count)
    except ValueError:
        continue

    if date != current_date:
        if current_date:
            for url, cnt in nlargest(5, entries, key=lambda x: x[1]):
                print(f"{current_date}\t{url}\t{cnt}")
        current_date = date
        entries = [(url, count)]
    else:
        entries.append((url, count))


if current_date:
    for url, cnt in nlargest(5, entries, key=lambda x: x[1]):
        print(f"{current_date}\t{url}\t{cnt}")
````

````python
data = spark.sparkContext.textFile("/user/student_user14/input_data/data_url.csv")

def extract_date(timestamp):
    return timestamp.split()[0]

# Get ((data, web-site), 1)
parsed_rdd = data.map(lambda line: line.split(';')) \
                    .map(lambda parts: ((extract_date(parts[1]), parts[0]), 1))


counts_rdd = parsed_rdd.reduceByKey(lambda a, b: a + b)

# Transform (data, (website, count)) 
date_website_count = counts_rdd.map(lambda x: (x[0][0], (x[0][1], x[1])))


grouped_by_date = date_website_count.groupByKey() \
                               .mapValues(lambda sites: sorted(sites, 
                                                             key=lambda x: -x[1]))

top5_by_date = grouped_by_date.mapValues(lambda sites: sites[:5])

results = top5_by_date.collect()


for date, sites in sorted(results, key=lambda x: x[0]):
    print(f"\nDate: {date}")
    print("Top-5:")
    for i, (site, count) in enumerate(sites, 1):
        print(f"{i}. {site} - {count}")

````

````python
#2. средняя зарплата в каждом подразделении
````

````python
data = spark.sparkContext.textFile('/user/student_user14/input_data/employees.csv')
````

````python
parsed_rdd = data.map(lambda line: line.split(',')) \
                    .map(lambda parts: (parts[2], float(parts[3])))

````

````python

zero_value = (0.0, 0)

def seq_op(accumulator, salary):
    return (accumulator[0] + salary, accumulator[1] + 1)

def comb_op(accum1, accum2):
    return (accum1[0] + accum2[0], accum1[1] + accum2[1])

avg_salary_rdd = parsed_rdd.aggregateByKey(zero_value, seq_op, comb_op) \
                          .mapValues(lambda x: x[0]/x[1])
````

````python
output_path = "/user/student_user14/output/avg_salary_results"

try:
    avg_salary_rdd.saveAsTextFile(output_path)
except Py4JJavaError as e:
    if "FileAlreadyExistsException" in str(e):
        hadoop = spark._jvm.org.apache.hadoop
        fs = hadoop.fs.FileSystem.get(hadoop.conf.Configuration())
        fs.delete(hadoop.fs.Path(output_path), True)  
        avg_salary_rdd.saveAsTextFile(output_path)
    else:
        raise e 
````
